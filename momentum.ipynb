```python
%load_ext autoreload
%autoreload 2
import pandas as pd
import plotly.graph_objects as go
import vectorbt as vbt

import data_helper

START_DATE = "2020-01-01"
END_DATE = "2022-01-01"
INITIAL_CASH = 3000.0
FEES = 0.001
FIXED_FEES = 1.00
SLIPPAGE = 0.0005
```

```python
close = data_helper.load_csvs()
```

```python
# --------------------------------------------------
# Buy-and-Hold simulieren
# --------------------------------------------------

portfolio = vbt.Portfolio.from_holding(
    close["ALV.DE"],
    init_cash=INITIAL_CASH,
    fees=0.0,
    fixed_fees=0.0,
    slippage=0.0,
    freq="1D"
)

# --------------------------------------------------
# Ergebnisse
# --------------------------------------------------

results = pd.DataFrame({
    "Endwert (€)": portfolio.value().iloc[-1],
    "Gesamtrendite (%)": portfolio.total_return() * 100,
    "Annualisierte Rendite (%)": portfolio.annualized_return() * 100,
    "Max. Drawdown (%)": portfolio.max_drawdown() * 100,
}, index=["ALV.DE"])

print("\nErgebnisse:")
print(results.round(2))
```

```python
values = data_helper.momentum_portfolio(
    close,
    start_date="2024-01-01",
    end_date="2026-01-01",
    lookback=126,
    topn=3,
    initial_cash=INITIAL_CASH,
    fees=FEES,
    fixed_fees=FIXED_FEES,
    slippage=SLIPPAGE
).value()

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=values.index,
    y=values,
    mode="lines",
))
fig.show()
```
