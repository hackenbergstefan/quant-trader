```python
universe = pytrading.data_helper.Universe()
```

```python
portfolio = pytrading.Portfolio(
    universe=universe,
    start_date="2015-01-01",
    end_date="2022-01-01",
    broker=pytrading.BrokerConfig(
        initial_cash=3000.0, fees=0.001, fixed_fees=1.00, slippage=0.0005
    ),
)
```

```python
pf = momentum.backtest(
    portfolio=portfolio,
    momentum=momentum.MomentumConfig(
        topn=10,
        min_avg_daily_turnover=1000000.0,
    )
)
values = pf.value()

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=values.index,
    y=values,
    mode="lines",
))
fig.show()
```
